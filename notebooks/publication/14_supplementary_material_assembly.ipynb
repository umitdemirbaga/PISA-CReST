```python
# Build Supplementary Table S1:
# Seven-fold education-system assignment and sample sizes

import pandas as pd
from pathlib import Path

# Load locked split files

fold_assignment = pd.read_csv(
    SPLITS_DIR / "country_fold_assignment_7fold.csv"
)

split_protocol = pd.read_csv(
    SPLITS_DIR / "country_split_protocol_7fold.csv"
)

# Build human-readable system lists

rows = []

for _, r in split_protocol.iterrows():

    outer_fold = int(r["outer_fold"])
    test_fold = int(r["test_country_fold"])
    val_fold = int(r["validation_country_fold"])

    test_systems = sorted(
        fold_assignment.loc[
            fold_assignment["country_fold"] == test_fold,
            "CNT"
        ].tolist()
    )

    val_systems = sorted(
        fold_assignment.loc[
            fold_assignment["country_fold"] == val_fold,
            "CNT"
        ].tolist()
    )

    # Training systems = all systems not in validation or test folds
    train_systems = sorted(
        fold_assignment.loc[
            ~fold_assignment["country_fold"].isin(
                [test_fold, val_fold]
            ),
            "CNT"
        ].tolist()
    )

    assert len(train_systems) == 45
    assert len(val_systems) == 9
    assert len(test_systems) == 9

    rows.append({
        "Outer fold": outer_fold,
        "Training systems": 45,
        "Validation systems": ", ".join(val_systems),
        "Test systems": ", ".join(test_systems),
        "Train students": int(r["n_train_students"]),
        "Validation students": int(r["n_validation_students"]),
        "Test students": int(r["n_test_students"]),
    })

table_s1 = pd.DataFrame(rows)

print("=" * 120)
print("SUPPLEMENTARY TABLE S1")
print("=" * 120)

display(table_s1)

# Global integrity checks

all_test_systems = []
all_validation_systems = []

for _, r in split_protocol.iterrows():

    test_fold = int(r["test_country_fold"])
    val_fold = int(r["validation_country_fold"])

    all_test_systems.extend(
        fold_assignment.loc[
            fold_assignment["country_fold"] == test_fold,
            "CNT"
        ].tolist()
    )

    all_validation_systems.extend(
        fold_assignment.loc[
            fold_assignment["country_fold"] == val_fold,
            "CNT"
        ].tolist()
    )

print("\n" + "=" * 120)
print("INTEGRITY CHECKS")
print("=" * 120)

print("Unique systems in dataset:",
      fold_assignment["CNT"].nunique())

print("Total test assignments:",
      len(all_test_systems))

print("Unique test systems:",
      len(set(all_test_systems)))

print("Every system exactly once as test:",
      len(all_test_systems) == 63
      and len(set(all_test_systems)) == 63)

print("Total validation assignments:",
      len(all_validation_systems))

print("Unique validation systems:",
      len(set(all_validation_systems)))

print("Every system exactly once as validation:",
      len(all_validation_systems) == 63
      and len(set(all_validation_systems)) == 63)

# Save CSV

csv_path = SUPP_TABLE_DIR / \
    "Table_S1_seven_fold_system_assignment.csv"

table_s1.to_csv(csv_path, index=False)

print("\nSaved CSV:")
print(csv_path)

# Generate final two-panel Supplementary Table S1

latex_lines = [
    r"\\begin{table*}[t]",
    r"\\caption{Seven-fold education-system assignment used for confirmatory cross-system evaluation.}",
    r"\\label{tab:supp_split_assignment}",
    r"\\centering",
    r"\\renewcommand{\\arraystretch}{1.15}",
    "",
    r"\\textbf{Panel A. Validation and test education systems}",
    r"\\medskip",
    r"\\resizebox{\\textwidth}{!}{%",
    r"\\begin{tabular}{@{}cll@{}}",
    r"\\toprule",
    r"\\textbf{Outer fold} & \\textbf{Validation systems} & \\textbf{Test systems} \\\\",
    r"\\midrule",
]

for _, r in table_s1.iterrows():
    latex_lines.append(
        f"{int(r['Outer fold'])} & "
        f"{r['Validation systems']} & "
        f"{r['Test systems']} \\\\"
    )

latex_lines += [
    r"\\bottomrule",
    r"\\end{tabular}%",
    r"}",
    r"\\vspace{0.9em}",
    "",
    r"\\textbf{Panel B. Sample sizes by outer fold}",
    r"\\medskip",
    r"\\small",
    r"\\begin{tabular}{@{}crrr@{}}",
    r"\\toprule",
    r"\\textbf{Outer fold} & \\textbf{Training $N$} & \\textbf{Validation $N$} & \\textbf{Test $N$} \\\\",
    r"\\midrule",
]

for _, r in table_s1.iterrows():
    latex_lines.append(
        f"{int(r['Outer fold'])} & "
        f"{int(r['Train students']):,} & "
        f"{int(r['Validation students']):,} & "
        f"{int(r['Test students']):,} \\\\"
    )

latex_lines += [
    r"\\bottomrule",
    r"\\end{tabular}",
    r"\\begin{flushleft}",
    r"\\footnotesize",
    r"\\textit{Note.} System labels are the three-letter identifiers used in the PISA 2022 public-use files. "
    r"For each outer fold, the listed test systems were completely unseen during model development, "
    r"the listed validation systems were used for model selection, and the remaining 45 education systems "
    r"formed the training set. Each of the 63 education systems served exactly once as an unseen test system "
    r"and exactly once as a validation system.",
    r"\\end{flushleft}",
    r"\\end{table*}",
]

latex_text = "\n".join(latex_lines)
tex_path = SUPP_TABLE_DIR / "Table_S1_seven_fold_system_assignment.tex"
tex_path.write_text(latex_text, encoding="utf-8")

print("Saved:", tex_path)
```

```python
# Build Supplementary Table S2
# Fold-level computational and reproducibility audit
# Parameter count is audited separately.

import pandas as pd

# Load locked audit files

training_summary = pd.read_csv(
    RESULTS_DIR / "crest_final_7fold_training_summary.csv"
)

training_time = pd.read_csv(
    RESULTS_DIR / "crest_final_7fold_training_time_audit.csv"
)

inference_time = pd.read_csv(
    RESULTS_DIR / "crest_final_7fold_inference_time_audit.csv"
)

# Merge fold-level information

s2 = (
    training_summary[
        [
            "outer_fold",
            "best_epoch",
            "best_validation_reconstruction",
            "epochs_run",
        ]
    ]
    .merge(
        training_time[
            [
                "outer_fold",
                "total_training_hours",
                "mean_epoch_seconds",
            ]
        ],
        on="outer_fold",
        how="left",
        validate="one_to_one"
    )
    .merge(
        inference_time[
            [
                "outer_fold",
                "checkpoint_epoch",
                "n_students",
                "latent_inference_seconds",
                "students_per_second",
                "milliseconds_per_student",
                "model_updated_on_test",
            ]
        ],
        on="outer_fold",
        how="left",
        validate="one_to_one"
    )
)

# Integrity checks

assert len(s2) == 7
assert not s2.isna().any().any(), "Missing values appeared after merge."
assert (s2["best_epoch"] == s2["checkpoint_epoch"]).all()
assert (~s2["model_updated_on_test"]).all()

# Publication-oriented table

table_s2 = pd.DataFrame({
    "Outer fold": s2["outer_fold"].astype(int),
    "Selected epoch": s2["best_epoch"].astype(int),
    "Epochs run": s2["epochs_run"].astype(int),
    "Validation reconstruction":
        s2["best_validation_reconstruction"],
    "Training time (h)":
        s2["total_training_hours"],
    "Mean epoch time (min)":
        s2["mean_epoch_seconds"] / 60,
    "Test N":
        s2["n_students"].astype(int),
    "Inference (ms/student)":
        s2["milliseconds_per_student"],
    "Throughput (students/s)":
        s2["students_per_second"],
})

print("=" * 120)
print("SUPPLEMENTARY TABLE S2")
print("=" * 120)

display(
    table_s2.round({
        "Validation reconstruction": 4,
        "Training time (h)": 3,
        "Mean epoch time (min)": 2,
        "Inference (ms/student)": 4,
        "Throughput (students/s)": 1,
    })
)

# Aggregate statistics

total_epochs = int(s2["epochs_run"].sum())
total_training_hours = s2["total_training_hours"].sum()

total_test_students = int(s2["n_students"].sum())
total_inference_seconds = s2["latent_inference_seconds"].sum()

pooled_ms_per_student = (
    total_inference_seconds / total_test_students
) * 1000

pooled_throughput = (
    total_test_students / total_inference_seconds
)

weighted_mean_epoch_seconds = (
    (s2["mean_epoch_seconds"] * s2["epochs_run"]).sum()
    / total_epochs
)

print("\n" + "=" * 120)
print("INTEGRITY / AGGREGATE CHECKS")
print("=" * 120)

print(
    "Selected epoch equals inference checkpoint epoch in all folds:",
    (s2["best_epoch"] == s2["checkpoint_epoch"]).all()
)

print(
    "All confirmatory test evaluations model-frozen:",
    (~s2["model_updated_on_test"]).all()
)

print(f"Total epochs executed: {total_epochs}")
print(f"Total final training time: {total_training_hours:.3f} h")
print(
    f"Mean epoch time: "
    f"{weighted_mean_epoch_seconds / 60:.2f} min"
)
print(f"Total held-out students: {total_test_students:,}")
print(
    f"Pooled latent inference: "
    f"{pooled_ms_per_student:.4f} ms/student"
)
print(
    f"Pooled throughput: "
    f"{pooled_throughput:.2f} students/s"
)

# Save CSV

csv_path = (
    SUPP_TABLE_DIR /
    "Table_S2_fold_level_computational_audit.csv"
)

table_s2.to_csv(csv_path, index=False)

print("\nSaved CSV:")
print(csv_path)

# Generate LaTeX

latex_lines = []

latex_lines.append(r"\begin{table*}[t]")
latex_lines.append(
    r"\caption{Fold-level computational and reproducibility "
    r"characteristics of the locked CReST confirmatory training "
    r"and inference protocol.}"
)
latex_lines.append(r"\label{tab:supp_computational_audit}")
latex_lines.append(r"\centering")
latex_lines.append(r"\scriptsize")
latex_lines.append(r"\renewcommand{\arraystretch}{1.10}")
latex_lines.append(r"\setlength{\tabcolsep}{4pt}")
latex_lines.append("")
latex_lines.append(r"\begin{tabular}{@{}ccccccccc@{}}")
latex_lines.append(r"\toprule")
latex_lines.append(
    r"\textbf{Fold} & "
    r"\textbf{Selected} & "
    r"\textbf{Epochs} & "
    r"\textbf{Validation} & "
    r"\textbf{Training} & "
    r"\textbf{Epoch time} & "
    r"\textbf{Test} & "
    r"\textbf{Inference} & "
    r"\textbf{Throughput} \\"
)
latex_lines.append(
    r" & "
    r"\textbf{epoch} & "
    r"\textbf{run} & "
    r"\textbf{reconstruction} & "
    r"\textbf{time (h)} & "
    r"\textbf{(min)} & "
    r"\textbf{$N$} & "
    r"\textbf{(ms/student)} & "
    r"\textbf{(students/s)} \\"
)
latex_lines.append(r"\midrule")

for _, r in table_s2.iterrows():
    latex_lines.append(
        f"{int(r['Outer fold'])} & "
        f"{int(r['Selected epoch'])} & "
        f"{int(r['Epochs run'])} & "
        f"{r['Validation reconstruction']:.4f} & "
        f"{r['Training time (h)']:.3f} & "
        f"{r['Mean epoch time (min)']:.2f} & "
        f"{int(r['Test N']):,} & "
        f"{r['Inference (ms/student)']:.4f} & "
        f"{r['Throughput (students/s)']:.1f} \\\\"
    )

latex_lines.append(r"\midrule")

latex_lines.append(
    f"\\multicolumn{{3}}{{l}}{{\\textbf{{Overall}}}} & "
    f"-- & "
    f"{total_training_hours:.2f} & "
    f"{weighted_mean_epoch_seconds / 60:.2f} & "
    f"{total_test_students:,} & "
    f"{pooled_ms_per_student:.3f} & "
    f"{pooled_throughput:.1f} \\\\"
)

latex_lines.append(r"\bottomrule")
latex_lines.append(r"\end{tabular}")
latex_lines.append("")
latex_lines.append(
    r"\begin{flushleft}"
    r"\footnotesize "
    r"\textit{Note.} Training time is the sum of the recorded "
    r"wall-clock epoch durations for all epochs actually executed "
    r"in each outer fold. Inference refers to frozen latent "
    r"representation extraction on the corresponding completely "
    r"held-out test systems. All training and confirmatory inference "
    r"were conducted on the same CPU-only workstation "
    r"(Intel Core i7-6700HQ, 2.60\,GHz; 4 physical / 8 logical "
    r"cores; approximately 24\,GB RAM), with no CUDA-enabled GPU. "
    r"No model parameters were updated during confirmatory test "
    r"evaluation."
    r"\end{flushleft}"
)

latex_lines.append(r"\end{table*}")

latex_text = "\n".join(latex_lines)

tex_path = (
    SUPP_TABLE_DIR /
    "Table_S2_fold_level_computational_audit.tex"
)

tex_path.write_text(latex_text, encoding="utf-8")

print("\nSaved LaTeX:")
print(tex_path)

print("\n" + "=" * 120)
print("LATEX PREVIEW")
print("=" * 120)
print(latex_text)
```

```python
# Identify state_dict tensors not represented in optimizer state

from pathlib import Path
from collections import Counter, defaultdict
import torch

BASE_DIR = Path.cwd()
CKPT_PATH = (
    BASE_DIR
    / "checkpoints"
    / "final_7fold"
    / "fold_0"
    / "crest_final_best.pt"
)

ckpt = torch.load(
    CKPT_PATH,
    map_location="cpu",
    weights_only=False
)

model_state = ckpt["model_state_dict"]
optimizer_state = ckpt["optimizer_state_dict"]

# Collect state_dict tensor signatures

state_entries = []

for key, tensor in model_state.items():
    if torch.is_tensor(tensor):
        state_entries.append({
            "key": key,
            "shape": tuple(tensor.shape),
            "numel": tensor.numel(),
            "dtype": str(tensor.dtype),
        })

# Collect optimizer parameter signatures from exp_avg

opt_shapes = []

for state in optimizer_state["state"].values():
    if "exp_avg" in state:
        t = state["exp_avg"]
        opt_shapes.append(
            (tuple(t.shape), t.numel())
        )

opt_counter = Counter(opt_shapes)

# Match state_dict tensors against optimizer tensor signatures
# using a multiset comparison

matched_counter = Counter()
candidate_nonoptimised = []

for entry in state_entries:

    sig = (entry["shape"], entry["numel"])

    if matched_counter[sig] < opt_counter[sig]:
        matched_counter[sig] += 1
    else:
        candidate_nonoptimised.append(entry)

print("=" * 100)
print("CANDIDATE STATE_DICT TENSORS NOT REPRESENTED IN OPTIMIZER STATE")
print("=" * 100)

total_extra = 0

for entry in candidate_nonoptimised:
    print(
        f"{entry['key']:<70} "
        f"shape={str(entry['shape']):<20} "
        f"numel={entry['numel']:<8} "
        f"dtype={entry['dtype']}"
    )
    total_extra += entry["numel"]

print("\n" + "=" * 100)
print("SUMMARY")
print("=" * 100)

print("Candidate tensors :", len(candidate_nonoptimised))
print("Candidate elements:", total_extra)
print("Expected tensors  :", 8)
print("Expected elements :", 626)

print(
    "\nExact match to audit discrepancy:",
    len(candidate_nonoptimised) == 8
    and total_extra == 626
)
```

```python
# Build Supplementary Table S5:
# Domain- and process-specific incremental information
# with 63-system bootstrap confidence intervals

import numpy as np
import pandas as pd

BOOTSTRAP_REPS = 50_000
BOOTSTRAP_SEED = 20270017

# Collapse A->B and B->A within each education system

domain_system = (
    domain_63
    .groupby(["domain", "country"], as_index=False)
    .agg(
        delta_rho=("delta_rho", "mean"),
        general_rho=("general_rho", "mean"),
        combined_rho=("combined_rho", "mean"),
    )
)

process_system = (
    process_63
    .groupby(["process", "country"], as_index=False)
    .agg(
        delta_rho=("delta_rho", "mean"),
        general_rho=("general_rho", "mean"),
        combined_rho=("combined_rho", "mean"),
    )
)

assert (domain_system.groupby("domain")["country"].nunique() == 63).all()
assert (process_system.groupby("process")["country"].nunique() == 63).all()


# System-level bootstrap helper

def system_bootstrap_summary(
    df,
    category_col,
    value_col="delta_rho",
    n_boot=50_000,
    seed=20270017,
):
    rng = np.random.default_rng(seed)

    rows = []

    for category, g in df.groupby(category_col, sort=False):

        values = g[value_col].to_numpy(dtype=float)
        n = len(values)

        estimate = values.mean()

        boot_idx = rng.integers(
            0,
            n,
            size=(n_boot, n)
        )

        boot_means = values[boot_idx].mean(axis=1)

        ci_low, ci_high = np.quantile(
            boot_means,
            [0.025, 0.975]
        )

        positive = int((values > 0).sum())

        rows.append({
            category_col: category,
            "systems": n,
            "system_equal_delta_rho": estimate,
            "ci_low": ci_low,
            "ci_high": ci_high,
            "median_delta_rho": np.median(values),
            "systems_positive": positive,
            "pct_systems_positive": positive / n * 100,
        })

    return pd.DataFrame(rows)


domain_boot = system_bootstrap_summary(
    domain_system,
    category_col="domain",
    n_boot=BOOTSTRAP_REPS,
    seed=BOOTSTRAP_SEED
)

process_boot = system_bootstrap_summary(
    process_system,
    category_col="process",
    n_boot=BOOTSTRAP_REPS,
    seed=BOOTSTRAP_SEED + 1
)


# Audit mean estimates against locked summary outputs

domain_locked = domain_summary[
    [
        "domain",
        "country_equal_delta_rho_mean",
    ]
].copy()

process_locked = process_summary[
    [
        "process",
        "country_equal_delta_rho_mean",
    ]
].copy()

domain_audit = domain_boot.merge(
    domain_locked,
    on="domain",
    how="left",
    validate="one_to_one"
)

process_audit = process_boot.merge(
    process_locked,
    on="process",
    how="left",
    validate="one_to_one"
)

domain_audit["mean_difference"] = (
    domain_audit["system_equal_delta_rho"]
    - domain_audit["country_equal_delta_rho_mean"]
)

process_audit["mean_difference"] = (
    process_audit["system_equal_delta_rho"]
    - process_audit["country_equal_delta_rho_mean"]
)


# Display audit

print("=" * 120)
print("DOMAIN AUDIT")
print("=" * 120)

display(
    domain_audit[
        [
            "domain",
            "system_equal_delta_rho",
            "country_equal_delta_rho_mean",
            "mean_difference",
            "ci_low",
            "ci_high",
            "systems_positive",
            "pct_systems_positive",
            "median_delta_rho",
        ]
    ].round(6)
)

print("\n" + "=" * 120)
print("PROCESS AUDIT")
print("=" * 120)

display(
    process_audit[
        [
            "process",
            "system_equal_delta_rho",
            "country_equal_delta_rho_mean",
            "mean_difference",
            "ci_low",
            "ci_high",
            "systems_positive",
            "pct_systems_positive",
            "median_delta_rho",
        ]
    ].round(6)
)

domain_match = np.allclose(
    domain_audit["system_equal_delta_rho"],
    domain_audit["country_equal_delta_rho_mean"],
    atol=1e-12
)

process_match = np.allclose(
    process_audit["system_equal_delta_rho"],
    process_audit["country_equal_delta_rho_mean"],
    atol=1e-12
)

print("\nMean estimates reproduce locked summaries:")
print("Domains :", domain_match)
print("Processes:", process_match)


# Publication ordering

domain_order = [
    "Social",
    "Scientific",
    "Written",
    "Visual",
]

process_order = [
    "GDI",
    "EII",
    "GCI",
]

domain_boot["order"] = domain_boot["domain"].map(
    {v: i for i, v in enumerate(domain_order)}
)

process_boot["order"] = process_boot["process"].map(
    {v: i for i, v in enumerate(process_order)}
)

domain_boot = (
    domain_boot
    .sort_values("order")
    .drop(columns="order")
    .reset_index(drop=True)
)

process_boot = (
    process_boot
    .sort_values("order")
    .drop(columns="order")
    .reset_index(drop=True)
)


# Save CSV outputs

domain_csv = (
    SUPP_TABLE_DIR /
    "Table_S5A_domain_specific_increment_bootstrap.csv"
)

process_csv = (
    SUPP_TABLE_DIR /
    "Table_S5B_process_specific_increment_bootstrap.csv"
)

domain_boot.to_csv(domain_csv, index=False)
process_boot.to_csv(process_csv, index=False)


# Generate LaTeX Table S5

latex_lines = []

latex_lines.append(r"\begin{table*}[t]")
latex_lines.append(
    r"\caption{Incremental hidden-target information contributed by "
    r"domain-specific and process-anchored CReST representations beyond "
    r"the general creative representation. Estimates give equal weight "
    r"to each of the 63 education systems.}"
)
latex_lines.append(r"\label{tab:supp_domain_process_increment}")
latex_lines.append(r"\centering")
latex_lines.append(r"\small")
latex_lines.append(r"\renewcommand{\arraystretch}{1.10}")
latex_lines.append("")

# Panel A
latex_lines.append(
    r"\textbf{Panel A. Domain-specific representations}"
)
latex_lines.append(r"\par\smallskip")
latex_lines.append(
    r"\begin{tabular}{@{}lrrrrr@{}}"
)
latex_lines.append(r"\toprule")
latex_lines.append(
    r"\textbf{Domain} & "
    r"\textbf{$\Delta\rho$} & "
    r"\textbf{95\% CI} & "
    r"\textbf{Positive systems} & "
    r"\textbf{Positive (\%)} & "
    r"\textbf{Median $\Delta\rho$} \\"
)
latex_lines.append(r"\midrule")

for _, r in domain_boot.iterrows():
    latex_lines.append(
        f"{r['domain']} & "
        f"{r['system_equal_delta_rho']:.4f} & "
        f"[{r['ci_low']:.4f}, {r['ci_high']:.4f}] & "
        f"{int(r['systems_positive'])}/63 & "
        f"{r['pct_systems_positive']:.1f} & "
        f"{r['median_delta_rho']:.4f} \\\\"
    )

latex_lines.append(r"\bottomrule")
latex_lines.append(r"\end{tabular}")

latex_lines.append(r"\vspace{1em}")

# Panel B
latex_lines.append(
    r"\textbf{Panel B. Process-anchored representations}"
)
latex_lines.append(r"\par\smallskip")
latex_lines.append(
    r"\begin{tabular}{@{}lrrrrr@{}}"
)
latex_lines.append(r"\toprule")
latex_lines.append(
    r"\textbf{Process} & "
    r"\textbf{$\Delta\rho$} & "
    r"\textbf{95\% CI} & "
    r"\textbf{Positive systems} & "
    r"\textbf{Positive (\%)} & "
    r"\textbf{Median $\Delta\rho$} \\"
)
latex_lines.append(r"\midrule")

for _, r in process_boot.iterrows():
    latex_lines.append(
        f"{r['process']} & "
        f"{r['system_equal_delta_rho']:.4f} & "
        f"[{r['ci_low']:.4f}, {r['ci_high']:.4f}] & "
        f"{int(r['systems_positive'])}/63 & "
        f"{r['pct_systems_positive']:.1f} & "
        f"{r['median_delta_rho']:.4f} \\\\"
    )

latex_lines.append(r"\bottomrule")
latex_lines.append(r"\end{tabular}")

latex_lines.append("")
latex_lines.append(
    r"\begin{flushleft}"
    r"\footnotesize "
    r"\textit{Note.} $\Delta\rho$ is the within-system difference in "
    r"Spearman correlation between the combined representation "
    r"(general plus domain-specific or process-anchored representation) "
    r"and the general creative representation alone when predicting "
    r"completely hidden target-item performance. The A$\rightarrow$B "
    r"and B$\rightarrow$A estimates were first averaged within each "
    r"education system. Reported means then assign equal weight to the "
    r"63 systems. Confidence intervals are percentile intervals from "
    f"{BOOTSTRAP_REPS:,} nonparametric system-level bootstrap "
    r"resamples. Process representations are described as "
    r"\emph{process-anchored} because the process tokens interact "
    r"through self-attention and should not be interpreted as pure "
    r"independent process factors."
    r"\end{flushleft}"
)

latex_lines.append(r"\end{table*}")

latex_text = "\n".join(latex_lines)

tex_path = (
    SUPP_TABLE_DIR /
    "Table_S5_domain_process_increment.tex"
)

tex_path.write_text(
    latex_text,
    encoding="utf-8"
)

print("\nSaved:")
print(domain_csv)
print(process_csv)
print(tex_path)

print("\n" + "=" * 120)
print("LATEX PREVIEW")
print("=" * 120)
print(latex_text)
```

```python
# Build Supplementary Table S6:
# Survey-weighted contextual associations and sensitivity analysis

import pandas as pd
import numpy as np

survey_dir = RESULTS_DIR / "survey_weighted_population"

# Load locked source files

context_bootstrap = pd.read_csv(
    survey_dir /
    "crest_population_context_system_bootstrap.csv"
)

context_summary = pd.read_csv(
    survey_dir /
    "crest_population_context_contrasts_summary.csv"
)

standardized_summary = pd.read_csv(
    survey_dir /
    "crest_population_standardized_effects_summary.csv"
)

adjusted_locked = pd.read_csv(
    survey_dir /
    "crest_population_adjusted_findings_LOCKED.csv"
)

adjusted_min30 = pd.read_csv(
    survey_dir /
    "crest_adjusted_associations_summary_min30.csv"
)

# Panel A:
# Unadjusted survey-weighted contrasts
# Contrast direction = Group B - Group A

panel_a = (
    context_summary[
        [
            "contrast",
            "group_A",
            "group_B",
            "n_systems",
            "country_equal_difference",
        ]
    ]
    .merge(
        context_bootstrap[
            [
                "contrast",
                "system_boot_ci_low",
                "system_boot_ci_high",
                "positive_systems",
                "negative_systems",
            ]
        ],
        on="contrast",
        how="left",
        validate="one_to_one"
    )
    .merge(
        standardized_summary[
            [
                "contrast",
                "mean_standardized_difference",
                "system_boot_ci_low",
                "system_boot_ci_high",
            ]
        ].rename(
            columns={
                "system_boot_ci_low": "std_ci_low",
                "system_boot_ci_high": "std_ci_high",
            }
        ),
        on="contrast",
        how="left",
        validate="one_to_one"
    )
)

panel_a["Direction"] = (
    panel_a["group_B"].astype(str)
    + " $-$ "
    + panel_a["group_A"].astype(str)
)

panel_a["positive_pct"] = (
    panel_a["positive_systems"]
    / panel_a["n_systems"]
    * 100
)

panel_a = panel_a[
    [
        "contrast",
        "Direction",
        "n_systems",
        "country_equal_difference",
        "system_boot_ci_low",
        "system_boot_ci_high",
        "mean_standardized_difference",
        "std_ci_low",
        "std_ci_high",
        "positive_systems",
        "negative_systems",
        "positive_pct",
    ]
]

# Panel B:
# Adjusted associations, primary >=30 and sensitivity >=50

panel_b = adjusted_locked.copy()

# Add positive-system percentage from the primary min30 analysis
panel_b = panel_b.merge(
    adjusted_min30[
        [
            "coefficient",
            "pct_positive_systems",
        ]
    ],
    on="coefficient",
    how="left",
    validate="one_to_one"
)

# More publication-friendly labels
coefficient_labels = {
    "HOMEPOS": "Home-resource index (HOMEPOS)",
    "Male": "Male",
    "Repeated grade": "Repeated grade",
    "Immigrant background": "Immigrant background",
    "Language mismatch": "Home/test language mismatch",
    "Age": "Age",
}

panel_b["Predictor"] = (
    panel_b["coefficient"]
    .map(coefficient_labels)
    .fillna(panel_b["coefficient"])
)

panel_b = panel_b[
    [
        "Predictor",
        "primary_n_systems",
        "primary_estimate",
        "primary_ci_low",
        "primary_ci_high",
        "pct_positive_systems",
        "sensitivity_n_systems",
        "sensitivity_estimate",
        "sensitivity_ci_low",
        "sensitivity_ci_high",
        "same_direction_supported_both",
        "interpretation",
    ]
]

# Display

print("=" * 120)
print("SUPPLEMENTARY TABLE S6 - PANEL A")
print("=" * 120)

display(
    panel_a.round({
        "country_equal_difference": 6,
        "system_boot_ci_low": 6,
        "system_boot_ci_high": 6,
        "mean_standardized_difference": 6,
        "std_ci_low": 6,
        "std_ci_high": 6,
        "positive_pct": 1,
    })
)

print("\n" + "=" * 120)
print("SUPPLEMENTARY TABLE S6 - PANEL B")
print("=" * 120)

display(
    panel_b.round({
        "primary_estimate": 6,
        "primary_ci_low": 6,
        "primary_ci_high": 6,
        "pct_positive_systems": 1,
        "sensitivity_estimate": 6,
        "sensitivity_ci_low": 6,
        "sensitivity_ci_high": 6,
    })
)

# Integrity checks against locked headline results

checks = {
    "Home resources": 0.105813,
    "Grade repetition": -0.101497,
    "Immigrant background": -0.011724,
    "Gender": -0.038623,
    "Home/test language": -0.033724,
}

print("\n" + "=" * 120)
print("INTEGRITY CHECKS")
print("=" * 120)

for contrast, expected in checks.items():
    observed = float(
        panel_a.loc[
            panel_a["contrast"] == contrast,
            "country_equal_difference"
        ].iloc[0]
    )

    print(
        f"{contrast:<25} "
        f"observed={observed:+.6f} "
        f"expected={expected:+.6f} "
        f"match={np.isclose(observed, expected, atol=1e-6)}"
    )

print(
    "\nPrimary adjusted systems:",
    sorted(panel_b["primary_n_systems"].unique())
)

print(
    "Sensitivity adjusted systems:",
    sorted(panel_b["sensitivity_n_systems"].unique())
)

# Save CSV files

panel_a_path = (
    SUPP_TABLE_DIR /
    "Table_S6A_unadjusted_context_associations.csv"
)

panel_b_path = (
    SUPP_TABLE_DIR /
    "Table_S6B_adjusted_context_associations.csv"
)

panel_a.to_csv(panel_a_path, index=False)
panel_b.to_csv(panel_b_path, index=False)

# Generate LaTeX

latex_lines = []

latex_lines.append(r"\begin{table*}[t]")
latex_lines.append(
    r"\caption{Survey-weighted contextual associations with the "
    r"CReST transfer score. Panel A reports unadjusted system-equal "
    r"contrasts and Panel B reports adjusted associations together "
    r"with the stricter support-threshold sensitivity analysis.}"
)
latex_lines.append(r"\label{tab:supp_context_associations}")
latex_lines.append(r"\centering")
latex_lines.append(r"\scriptsize")
latex_lines.append(r"\renewcommand{\arraystretch}{1.10}")
latex_lines.append("")

# Panel A

latex_lines.append(
    r"\textbf{Panel A. Unadjusted survey-weighted contextual contrasts}"
)
latex_lines.append(r"\par\smallskip")

latex_lines.append(
    r"\begin{tabular}{@{}lrrrrrr@{}}"
)

latex_lines.append(r"\toprule")

latex_lines.append(
    r"\textbf{Contrast} & "
    r"\textbf{Systems} & "
    r"\textbf{Difference} & "
    r"\textbf{95\% CI} & "
    r"\textbf{Std. difference} & "
    r"\textbf{95\% CI} & "
    r"\textbf{Positive systems} \\"
)

latex_lines.append(r"\midrule")

for _, r in panel_a.iterrows():

    latex_lines.append(
        f"{r['contrast']} & "
        f"{int(r['n_systems'])} & "
        f"{r['country_equal_difference']:+.4f} & "
        f"[{r['system_boot_ci_low']:.4f}, "
        f"{r['system_boot_ci_high']:.4f}] & "
        f"{r['mean_standardized_difference']:+.3f} & "
        f"[{r['std_ci_low']:.3f}, "
        f"{r['std_ci_high']:.3f}] & "
        f"{int(r['positive_systems'])}/"
        f"{int(r['n_systems'])} \\\\"
    )

latex_lines.append(r"\bottomrule")
latex_lines.append(r"\end{tabular}")

latex_lines.append(r"\vspace{1em}")

# Panel B

latex_lines.append(
    r"\textbf{Panel B. Adjusted associations and support-threshold sensitivity}"
)

latex_lines.append(r"\par\smallskip")

latex_lines.append(
    r"\begin{tabular}{@{}lrrrrrr@{}}"
)

latex_lines.append(r"\toprule")

latex_lines.append(
    r"\textbf{Predictor} & "
    r"\textbf{$N_{\mathrm{sys}}$} & "
    r"\textbf{Primary estimate} & "
    r"\textbf{95\% CI} & "
    r"\textbf{Positive (\%)} & "
    r"\textbf{Sensitivity estimate} & "
    r"\textbf{95\% CI} \\"
)

latex_lines.append(r"\midrule")

for _, r in panel_b.iterrows():

    latex_lines.append(
        f"{r['Predictor']} & "
        f"{int(r['primary_n_systems'])} & "
        f"{r['primary_estimate']:+.4f} & "
        f"[{r['primary_ci_low']:.4f}, "
        f"{r['primary_ci_high']:.4f}] & "
        f"{r['pct_positive_systems']:.1f} & "
        f"{r['sensitivity_estimate']:+.4f} & "
        f"[{r['sensitivity_ci_low']:.4f}, "
        f"{r['sensitivity_ci_high']:.4f}] \\\\"
    )

latex_lines.append(r"\bottomrule")
latex_lines.append(r"\end{tabular}")

latex_lines.append("")

latex_lines.append(
    r"\begin{flushleft}"
    r"\footnotesize "
    r"\textit{Note.} The CReST transfer score is the average of the "
    r"A$\rightarrow$B and B$\rightarrow$A disjoint-item transfer "
    r"scores and is not an official PISA plausible value or proficiency "
    r"score. In Panel A, each contrast is defined as Group B minus "
    r"Group A: Q4 minus Q1 for home resources, repeated minus never "
    r"repeated, immigrant-background minus native, male minus female, "
    r"and home/test-language mismatch minus match. Estimates give equal "
    r"weight to education systems. Confidence intervals are "
    r"nonparametric system-bootstrap intervals. Standardised differences "
    r"are reported as a scale-free sensitivity description. In Panel B, "
    r"the primary adjusted analysis required a minimum supported sample "
    r"of 30 per included group or predictor configuration and retained "
    r"33 education systems; the sensitivity analysis increased this "
    r"threshold to 50 and retained 26 systems. Panel B confidence "
    r"intervals are also system-bootstrap intervals. Associations are "
    r"descriptive and should not be interpreted causally."
    r"\end{flushleft}"
)

latex_lines.append(r"\end{table*}")

latex_text = "\n".join(latex_lines)

tex_path = (
    SUPP_TABLE_DIR /
    "Table_S6_context_associations.tex"
)

tex_path.write_text(
    latex_text,
    encoding="utf-8"
)

print("\nSaved:")
print(panel_a_path)
print(panel_b_path)
print(tex_path)

print("\n" + "=" * 120)
print("LATEX PREVIEW")
print("=" * 120)
print(latex_text)
```
