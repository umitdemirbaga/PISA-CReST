# CReST: Manuscript Results Assembly

Assembles aggregate manuscript-result tables and claim/evidence summaries from final analysis artifacts. No student-level data are exported.

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
# Inventory all manuscript-relevant result artifacts

from pathlib import Path
import pandas as pd


BASE_DIR = Path.cwd()

RESULTS_DIR = BASE_DIR / "results"
FIG_DIR = BASE_DIR / "figures"
PROCESSED_DIR = BASE_DIR / "processed"
CHECKPOINT_DIR = BASE_DIR / "checkpoints"


MANUSCRIPT_DIR = (
    RESULTS_DIR
    / "manuscript_assembly"
)

MANUSCRIPT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Collect analysis outputs

extensions = {
    ".csv",
    ".json",
    ".npy",
    ".npz",
    ".pkl",
    ".pdf",
    ".png"
}


records = []


for root_name, root_dir in [
    ("results", RESULTS_DIR),
    ("figures", FIG_DIR),
    ("processed", PROCESSED_DIR)
]:

    if not root_dir.exists():
        continue


    for path in root_dir.rglob("*"):

        if (
            path.is_file()
            and
            path.suffix.lower()
            in extensions
        ):

            records.append(
                {
                    "root":
                        root_name,

                    "relative_path":
                        str(
                            path.relative_to(
                                BASE_DIR
                            )
                        ),

                    "filename":
                        path.name,

                    "extension":
                        path.suffix.lower(),

                    "size_mb":
                        path.stat().st_size
                        / (1024 ** 2)
                }
            )


artifact_inventory = pd.DataFrame(
    records
)


# Manuscript-relevant keyword subset

KEYWORDS = [
    "final",
    "test",
    "domain",
    "process",
    "signal",
    "context",
    "survey",
    "population",
    "pisa",
    "profile",
    "bootstrap",
    "sensitivity",
    "robust",
    "fold",
    "training",
    "alignment"
]


pattern = "|".join(
    KEYWORDS
)


relevant_inventory = (
    artifact_inventory[
        artifact_inventory[
            "relative_path"
        ]
        .str.lower()
        .str.contains(
            pattern,
            regex=True
        )
    ]
    .sort_values(
        "relative_path"
    )
    .reset_index(
        drop=True
    )
)


relevant_inventory.to_csv(
    MANUSCRIPT_DIR
    / "manuscript_artifact_inventory.csv",
    index=False
)


print(
    "Total artifact files:",
    len(
        artifact_inventory
    )
)

print(
    "Potentially manuscript-relevant:",
    len(
        relevant_inventory
    )
)


print(
    "\nMANUSCRIPT-RELEVANT FILES"
)

print(
    "=" * 100
)

print(
    relevant_inventory[
        [
            "relative_path",
            "size_mb"
        ]
    ]
    .to_string(
        index=False
    )
)
```
